## Préparation
Exécutez cette cellule une fois : elle importe `re` et définit la fonction `verifier`.

```python
import re

def verifier(nom, obtenu, attendu):
    """Affiche ✅ si le résultat est celui attendu, ❌ sinon."""
    if obtenu == attendu:
        print(f"✅ {nom}")
    else:
        print(f"❌ {nom}\n   obtenu  : {obtenu!r}\n   attendu : {attendu!r}")
```

## Exercice 1 ★ · Les nombres entiers
Écrire `nombres(texte)` qui renvoie la liste de tous les nombres entiers du texte.

```python
def nombres(texte):
    motif = r"\d+"
    return re.findall(motif, texte)
```

```python
verifier("nombres : test 1", nombres("J'ai 3 chats, 12 poissons et 105 fourmis."), ['3', '12', '105'])
verifier("nombres : test 2", nombres('Aucun chiffre ici.'), [])
```

## Exercice 2 ★ · Les années
Écrire `annees(texte)` qui renvoie les nombres d'exactement 4 chiffres (pas les morceaux d'un nombre plus long).

```python
def annees(texte):
    motif = r"(?<!\d)\d{4}(?!\d)"
    return re.findall(motif, texte)
```

```python
verifier("annees : test 1", annees('En 1999, puis en 2024 ; code 20245.'), ['1999', '2024'])
```

## Exercice 3 ★ · Les hashtags
Écrire `hashtags(texte)` qui renvoie les hashtags (un # suivi d'au moins un caractère de mot).

```python
def hashtags(texte):
    motif = r"#[\w]+"
    return re.findall(motif, texte)
```

```python
verifier("hashtags : test 1", hashtags('Vive #TALN et #ia2027 ! # seul'), ['#TALN', '#ia2027'])
```

## Exercice 4 ★ · Les mots qui commencent par une majuscule
Écrire `noms_propres(texte)` qui renvoie les mots commençant par une majuscule, **accents compris** (É, À…).

```python
def noms_propres(texte):
    motif = r"(?<![A-Za-zÀ-ÖØ-Þà-öø-ÿ])[A-ZÀ-ÖØ-Þ][A-Za-zÀ-ÖØ-Þà-öø-ÿ]*"
    return re.findall(motif, texte)
```

```python
verifier("noms_propres : test 1", noms_propres('Été à Fès, Rabat et Oujda.'), ['Été', 'Fès', 'Rabat', 'Oujda'])
```

## Exercice 5 ★★ · Les mots longs
Écrire `mots_longs(texte)` qui renvoie les mots de 7 caractères ou plus.

```python
def mots_longs(texte):
    motif = r"\b\w{7,}\b"
    return re.findall(motif, texte)
```

```python
verifier("mots_longs : test 1", mots_longs('Le traitement automatique du langage'), ['traitement', 'automatique', 'langage'])
```

## Exercice 6 ★★ · Les adresses e-mail
Écrire `emails(texte)` qui renvoie les adresses e-mail simples (nom@domaine.extension). Attention au point final de la phrase !

```python
def emails(texte):
    motif = r"\b[\w.]+@[\w.-]+\.[A-Za-z]{2,}\b"
    return re.findall(motif, texte)
```

```python
verifier("emails : test 1", emails('Écrire à sara.b@emsi.ma ou à ali_99@gmail.com.'), ['sara.b@emsi.ma', 'ali_99@gmail.com'])
```

## Exercice 7 ★★ · Les prix en dirhams
Écrire `prix(texte)` qui renvoie les prix suivis de « DH », avec ou sans espace, avec ou sans décimales.

```python
def prix(texte):
    motif = r"\d+(?:,\d+)?\s*DH"
    return re.findall(motif, texte)
```

```python
verifier("prix : test 1", prix('Café 12 DH, menu 89,50DH et pain 1 DH.'), ['12 DH', '89,50DH', '1 DH'])
```

## Exercice 8 ★★ · Les espaces multiples
Écrire `espaces(texte)` qui remplace toute suite de blancs (espaces, tabulations, retours à la ligne) par un seul espace et retire les blancs au début et à la fin.

```python
def espaces(texte):
    return re.sub(r"\s+", " ", texte).strip()
```

```python
verifier("espaces : test 1", espaces('  Bonjour \t  tout\nle monde  '), 'Bonjour tout le monde')
```

## Exercice 9 ★★ · Les balises HTML
Écrire `sans_balises(texte)` qui supprime les balises HTML. Pourquoi `<.+>` ne convient-il pas ?

```python
def sans_balises(texte):
    return re.sub(r"<[^>]+>", "", texte)
```

```python
verifier("sans_balises : test 1", sans_balises('<p>Très <b>bon</b> produit</p>'), 'Très bon produit')
```

## Exercice 10 ★★★ · Les heures
Écrire `heures(texte)` qui renvoie les heures au format « 9h05 », « 14h30 » ou « 14:30 », sans prendre de morceaux de nombres plus longs.

```python
def heures(texte):
    motif = r"(?<!\d)(?:[01]?\d|2[0-3])(?:h|:)[0-5]\d(?!\d)"
    return re.findall(motif, texte)
```

```python
verifier("heures : test 1", heures('RDV à 9h05, puis 14:30 ; pas 123h456 ni 2024.'), ['9h05', '14:30'])
```

## Exercice 11 ★★★ · Les lignes d'une liste
Écrire `items(texte)` qui renvoie les lignes qui commencent par « - » (tiret puis espace). Il faut une option !

```python
def items(texte):
    motif = r"^- .*"
    option = re.MULTILINE
    return re.findall(motif, texte, flags=option)
```

```python
verifier("items : test 1", items('- pain\n- lait\nnote : vite\n- dattes'), ['- pain', '- lait', '- dattes'])
```

## Exercice 12 ★★★ · Compter un mot entier
Écrire `compter_chat(texte)` qui compte le mot « chat » en entier (pas « chaton », pas « achat »), quelle que soit la casse.

```python
def compter_chat(texte):
    motif = r"\bchat\b"
    option = re.IGNORECASE
    return len(re.findall(motif, texte, flags=option))
```

```python
verifier("compter_chat : test 1", compter_chat('Chat, chaton, achat, CHAT et chat.'), 3)
```

## Pour aller plus loin ★★★
Écrire `mobiles(texte)` qui renvoie les numéros de mobile marocains de 10 chiffres collés, commençant par 06 ou 07.

```python
def mobiles(texte):
    motif = r"(?<!\d)(?:06|07)\d{8}(?!\d)"
    return re.findall(motif, texte)
```

```python
verifier("mobiles", mobiles("Appelez le 0612345678 ou le 0798765432, pas le 0512345678 ni 06123456789."), ['0612345678', '0798765432'])
```
